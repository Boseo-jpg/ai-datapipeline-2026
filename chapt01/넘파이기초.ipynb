### Python 리스트(배열) 과 NumPy 배열 비교

- 파이썬에 기존 리스트(배열) 자료구조 존재. 속도적면에서 NumPy가 우세

#### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격의 돈 배열
discount_prices = []

for price in prices:
    discount_prices.append(price * 0.9) # 10% 할인

discount_prices # 노트북 코드쉘에서 하나만 출력하려면 print() 생략 가능
````

#### 넘파이로 배열작업

````python
np_prices = np.array(prices) # 넘파이 배열로 변경
np_discount_prices = np_prices * 0.9

np_discount_prices
````

#### NumPy 데이터타입

- int8 : -128 ~ 127 (총 256 - 2의 8승)
- int16 : -32768 ~ 32767 (총 65536 - 2의 16승)
- int32 : 32비트 정수 (2의 32승)
- int64 : 64비트 정수 (2의 64승)
- uint8 : 음수 없는 정수 (0 ~ 255)
- uint16 : 음수 없는 정수 (0~ 65535)
- float16 : 실수
- float32 : 32비트 실수
- float64 : 64비트 실수 (NumPy 기본 실수)
- complex64 : 64비트 복소수
- complex128 : 128비트 복소수
- bool : True/False
- str_ : 문자열
- object_ : 객체

#### 기본 통계함수

- 평균, 최대/최소값, 중앙값, 표준편차 등

````python
scores = np.array([80, 95, 70, 60, 100, 85])
````

````python
# 합

np.sum(scores)
````

````python
# 평균

np.mean(scores)
````

````python
# 최대값

np.max(scores)
````

````python
# 최소값

np.min(scores)
````

````python
# 중앙값

np.median(scores)
````

````python
# 표준편차

np.std(scores)
````

#### 조건 필터링

- SQL WHERE 절과 동일, 조건에 맞는 데이터만 골라내기

````python
np_prices
````

````python
np_prices >= 20000
````

````python
condition = np_prices >= 20000
condition
````

````python
np_prices[condition]
````

````python
np_prices[np_prices >= 20000]
````

````python
datas = np.array([10, -2, 80, 999, 30, 120])
datas
````

````python
datas[(datas >= 0) & (datas <= 100)]
````
