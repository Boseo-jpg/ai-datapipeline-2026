````python
import pandas as pd
````

````python
# DataFrame을 직접 생성할때 사용. 
# 파이썬 딕셔너리 사용
raw_data = {
    'name' : ['홍길동', '김철수', '이영희', '박민수', '성유고', '애슐리'],
    'age' : [20, 21, 22, 23, 30, 29],
    'major' : ['데이터분석', '웹개발', '인공지능', '임베디드', '인공지능', '웹개발'],
    'score' : [85, 90, 95, 80, 90, 100]
}

raw_data
````

````python
df = pd.DataFrame(raw_data)

df
````

````python
# 숫자타입 필드만 통계 가능

df.describe()
````

#### 컬럼(열) 선택

````python
# DataFrame에서 필요한 컬럼만 확인 또는 할당(복사)

df['name']
````

````python
# 컬럼 선택의 편의성을 위해 추가된 방식. 위와 동일

df.name
````

#### 로우(행) 선택

````python
# 인덱스 이름을 기준

df.loc[0]
````

````python
# 위치번호 기준. 인덱스를 다른 값으로 바꿨을 때 사용하는 방법

df.iloc[0]
````

````python
df.iloc[0:2]
````

#### 행선택, 인덱스행(loc)과 위치행값(iloc) 차이

````python
df2 = pd.DataFrame(raw_data,
                   index=['S0001', 'S0003', 'S0005', 'S0007', 'S0009', 'S0011'])

df2
````

````python
df2.loc['S0001':'S0006']
````

````python
df2.iloc[1:4]
````

- loc는 부여받은 인덱스로 행선택, 슬라이싱
- iloc는 원래부터의 위치값(파이썬 배열의 인덱스)와 동일

#### 행과 열 함께 선택

````python
df.loc[0, 'name']
````

````python
# 인덱스를 따로 부여한 DF에서 행과 열 선택

df2.loc['S0001':'S0004', ['name', 'age']]
````

#### 조건조회

- SQL WHERE 절과 동일

````python
df['score'] >= 90
````

````python
df[df['score'] >= 90]
````

- NumPy와 동일한 조건 연산
- 파이썬 : and, or, not 사용
- NumPy, Pandas : &, |, ~ 사용

````python
# 점수가 90점 초과이고 나이가 21살 이상인 학생 데이터 참/거짓 여부

(df['score'] > 90) & (df['age'] >= 21)
````

````python
df[(df['score'] > 90) & (df['age'] >= 21)]
````

````python
# 점수가 90점 초과이거나 나이가 21살 이상 둘 중 하나라도 참인 데이터 조회

df[(df['score'] > 90) | (df['age'] >= 21)]
````

#### 컬럼 추가

#### 컬럼 수정

- 기존 컬럼 수정

````python
df['score_bonus'] = df['score_bonus'] - 2

df
````

````python
# 행 선택에서 score_bonus 점수가 100점 초과하는 행만 선택 후 score_bonus 값을 100으로 수정

df.loc[df['score_bonus'] > 100, 'score_bonus'] = 100

df
````

### 실습

- 상품데이터 활용

````python
data = {
    'product' : ['키보드', '마우스', '모니터', 'USB', '노트북'],
    'category' : ['주변기기', '주변기기', '디스플레이', '저장장치', '컴퓨터'],
    'price' : [30000, 15000, 250000, 12000, 1200000],
    'stock' : [10, 30, 5, 50, 3]
}

df_products = pd.DataFrame(data)
````

1. 전체 데이터 출력
2. product, price 컬럼 출력
3. 가격이 50000원 이상인 상품만 출력
4. 재고가 10개 이상인 상품만 출력
5. total_value 컬럼 추가, 계산식은 price * stock
6. 전체 재고 가치의 합계 출력
7. 카테고리가 주변기기인 상품만 출력

````python
# 1. 전체 데이터 출력

df_products
````

````python
# 2. product, price 컬럼 출력

df_products[['product', 'price']]
````

````python
# 3. 가격이 50000원 이상인 상품만 출력

df_products[df_products['price'] >= 50000]
````

````python
# 4. 재고가 10개 이상인 상품만 출력

df_products[df_products['stock'] >= 10]
````

````python
# 5. total_value 컬럼 추가, 계산식은 price * stock

df_products['total_value'] = df_products['price'] * df_products['stock']

df_products
````

````python
# 6. 전체 재고의 합계 출력

df_products['total_value'].sum()
````

````python
# 7. 카테고리가 주변기기인 상품만 출력

df_products[df_products['category'] == '주변기기']
````
